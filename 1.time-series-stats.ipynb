```python
import numpy as np
import matplotlib.pyplot as plt
print("Setup successful!")
plt.plot([1, 2, 3], [1, 4, 9])
plt.show()
```

```python
sampling_rate = 1000 #hertz
duration = 2 #seconds
time = np.linspace(0,duration, sampling_rate * duration)
print(f"Time points: {len(time)}")
```

```python
frequency = 8 #hertz
clean_signal = np.sin(2 * np.pi * frequency * time)
noise = np.random.normal(0, 0.5, len(time))
noisy_signal = clean_signal + noise

window_size = 25
kernel = np.ones(window_size) / window_size
smoothed_signal = np.convolve(noisy_signal, kernel, mode ="same")
```

```python
plt.figure(figsize=(12, 4))
plt.plot(time, noisy_signal, label="Noisy Signal", alpha=0.7)
plt.plot(time, smoothed_signal, label = "Smoothed Signal", linewidth = 2)
plt.plot(time, clean_signal, label="Clean Signal", linewidth=2)
plt.xlabel("Time (s)")
plt.ylabel("Signal Amplitude")
plt.title("Simulated Neural Signal")
plt.legend()
plt.show()
```

```python
mean = np.mean(noisy_signal)
std = np.std(noisy_signal)
variance = np.var(noisy_signal)

print(f"Mean: {mean:.3f}")
print(f"Standard Deviation: {std:.3f}")
print(f"Variance: {variance:.3f}")
```

```python
plt.figure(figsize=(6, 4))
plt.hist(noisy_signal, bins=50, alpha=0.7)
plt.xlabel("Signal Amplitude")
plt.ylabel("Count")
plt.title("Distribution of Neural Signal Amplitudes")
plt.show()
```

```python
signal_A = clean_signal + np.random.normal(0, 0.4, len(time))
signal_B = clean_signal + np.random.normal(0, 0.8, len(time))


mean_A = np.mean(signal_A)
mean_B = np.mean(signal_B)

std_A = np.std(signal_A)
std_B = np.std(signal_B)

print(f"Condition A — Mean: {mean_A:.3f}, Standard Deviation: {std_A:.3f}")
print(f"Condition B — Mean: {mean_B:.3f}, Standard Deviation: {std_B:.3f}")
```
